```python
from torchtext.data.utils import get_tokenizer
from torchtext.vocab import build_vocab_from_iterator

tokenizador = get_tokenizer("basic_english")
train_iter = iter(dataset["train"])

def yield_tokens(data_iter):
  for fila in data_iter:
    yield tokenizador(fila['content'])

vocab = build_vocab_from_iterator(yield_tokens(train_iter), specials=["<unk>"])
vocab.set_default_index(vocab["<unk>"])
```

```python
vocab(tokenizador("Hellos how are you? I am a platzi student"))
```

```python
texto_pipeline = lambda x: vocab(tokenizador(x))
label_pipeline = lambda x: int(x)
```

```python
texto_pipeline("Hellos I am Carlos")
```

```python
label_pipeline("20")
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

def callable_batch(batch):
    label_list, text_list, offsets = [], [], [0]

    for item in batch:
        # Extraemos los datos usando las llaves del diccionario de Hugging Face
        _label = item['label']
        _text = item['content']

        label_list.append(label_pipeline(_label))
        processed_text = torch.tensor(texto_pipeline(_text), dtype=torch.int64)
        text_list.append(processed_text)
        offsets.append(processed_text.size(0))

    label_list = torch.tensor(label_list, dtype=torch.int64)
    offsets = torch.tensor(offsets[:-1]).cumsum(dim=0)
    text_list = torch.cat(text_list)

    return label_list.to(device), text_list.to(device), offsets.to(device)
```

```python
from torch.utils.data import DataLoader

train_iter = iter(dataset["train"])
dataloder = DataLoader(train_iter,batch_size=8,shuffle=False, collate_fn=callable_batch)
```

```python
from torch import nn
import torch.nn.functional as F
```

```python
class ModeloClasificacionTexto(nn.Module):
  def __init__(self, vocab_size, embed_dim, num_class):
    super(ModeloClasificacionTexto, self).__init__()

    # Capa de incrustracion (embedding)
    self.embedding = nn.EmbeddingBag(vocab_size, embed_dim)

    # Capa de normalizacion por lotes
    self.bn1 = nn.BatchNorm1d(embed_dim, num_class)

    #Capa completamente conectada (fully connected)
    self.fc = nn.Linear(embed_dim, num_class)

  def forward(self, text, offsets):

    # Incrustar el texto (embed the text)
    embedded = self.embedding(text, offsets)

    # Aplicar la normalizacion por lotes (apply batch normalization)
    embedded_norm = self.bn1(embedded)

    # Alpicar la funcion ReLu (apply the ReLu activation function)
    embedded_activated = F.relu(embedded_norm)

    # devolver las probabilidades de clase (output class probabilites)
    return self.fc(embedded_activated)
```

```python
train_iter = dataset['train']
num_class = len(set([fila['label'] for fila in train_iter]))
vocab_size = len(vocab)
embedding_size = 100
```

```python
modelo = ModeloClasificacionTexto(vocab_size= vocab_size, embed_dim=embedding_size, num_class=num_class).to(device)
```

```python
#arquitectura
#print(modelo)

# Numero de prarametros entrenables en nuestro modelo

def count_parameters(model):
  return sum(p.numel()for p in model.parameters() if p.requires_grad)

print(f"El modelo tiene {count_parameters(modelo):,} parametros entrenables")

```

```python
def entrena(dataloder):
  # colacar el modelo en formato de entrenamiento
  modelo.train()

  #inicializar el accuracy, count y loss para cada epoch
  epoch_acc = 0
  epoch_loss = 0
  total_count = 0

  for idx, (label, text, offsets) in enumerate(dataloder):

    # reestablecer los gradientes despues de cada batch
    optimizer.zero_grad()
    # obtener predicciones del modelo
    prediccion = modelo(text, offsets)
    # obtener la perdida
    loss = criterio(prediccion, label)
    # backpropage la perdida y calcular los gradiantes
    loss.backward()
    #obtener la accuracy
    acc= (prediccion.argmax(dim=1) == label).sum()
    #evitar que los gradiantes sena demasiado grandes
    torch.nn.utils.clip_grad_norm_(modelo.parameters(), 0.1)
    # Actualizar los pesos
    optimizer.step()
    # llevamos el conteo de la perdida y el accuracy para esta epoch
    epoch_acc += acc.item()
    epoch_loss += loss.item()
    total_count += label.size(0)

    if idx % 500 ==0 and idx > 0:
      print(f"epcoa{epoch} | {idx}/{len(dataloder)} batches |  perdida {epoch_loss/total_count} | accuracy {epoch_acc/total_count}")
  return epoch_acc/total_count, epoch_loss/total_count


```

```python
def evalua(dataloder):
  modelo.eval()
  epoch_acc = 0
  epoch_loss = 0
  total_count = 0

  with torch.no_grad():
    for idx, (label, text, offsets) in enumerate(dataloder):
      #obtener la etiqueta predecida
      prediccion = modelo(text, offsets)
      loss = criterio(prediccion, label)
      acc = (prediccion.argmax(1) == label).sum()
      #llevamos el conteo de la perdida en el accuracy para esta epoch
      epoch_loss += loss.item()
      epoch_acc += acc.item()
      total_count += label.size(0)

  return epoch_acc/total_count, epoch_loss/total_count
```

```python
EPOCHS = 3
TASA_APRENDIZAJE = 0.2
BATCH_TAMANO = 64
```

```python
criterio = torch.nn.CrossEntropyLoss()
optimizer = torch.optim.SGD(modelo.parameters(), lr =TASA_APRENDIZAJE)
```

```python
from torch.utils.data.dataset import random_split
from torch.utils.data import DataLoader

# obtener el trainset y testset

train_dataset = dataset['train']
test_dataset = dataset['test']

#Entrenamos el modelo con el 95& de los datos del train set
num_train = int(len(train_dataset) * 0.95)
num_valid = len(train_dataset) - num_train

# Creamos un dataset de validacion con el 5% del trainset
split_train_, split_valid_ = random_split(train_dataset, [num_train, len(train_dataset) - num_train])

#creamos dataloder listos para ingrear a nuestro modelo

train_dataloader = DataLoader(split_train_, batch_size=BATCH_TAMANO, shuffle=True, collate_fn=callable_batch)
valid_dataloader = DataLoader(split_valid_, batch_size=BATCH_TAMANO, shuffle=True, collate_fn=callable_batch)
test_dataloader = DataLoader(test_dataset, batch_size=BATCH_TAMANO, shuffle=True, collate_fn=callable_batch)

print(f"Datos de entrenamiento: {len(split_train_)}")
print(f"Datos de validación: {len(split_valid_)}")
print(f"Datos de prueba: {len(test_dataset)}")
```

```python
#obtener la mejor perdida
major_loss_validation = float('inf')

# Entrenamos
for epoch in range(1, EPOCHS + 1):

  #entrenamiento
  entrenamiento_acc, entrenamiento_loss = entrena(train_dataloader)

  #Validacion
  validacion_acc, validacion_loss = evalua(valid_dataloader)

  #guardar el mejor madelo
  if validacion_loss < major_loss_validation:
    best_valid_loss = validacion_loss
    torch.save(modelo.state_dict(), "mejores_guardados.pt")
```

```python
test_acc, test_loss = evalua(test_dataloader)

print(f"Accuracy del test datase: {test_acc}")
print(f"Perdida del test dataset: {test_loss}")
```
