```python
import torch
from datasets import load_dataset
```

```python
dataset = load_dataset("fancyzhx/dbpedia_14")
```

```python
train_iter = iter(dataset['train'])
```

```python
next(train_iter)
```

```python
from torchtext.data.utils import get_tokenizer
from torchtext.vocab import build_vocab_from_iterator

tokenizador = get_tokenizer('basic_english')
train_iter = iter(dataset['train'])

def yield_tokens(data_iter):
    for item in data_iter:
      yield tokenizador(item['content'])

vocab = build_vocab_from_iterator(yield_tokens(train_iter), specials=["<unk>"])
vocab.set_default_index(vocab["<unk>"])
```

```python
vocab(tokenizador("Hello how are you? I am a platzi student"))
```

```python
texto_pipeline = lambda x: vocab(tokenizador(x))
label_pipeline = lambda x: int(x) - 1
```

```python
texto_pipeline("Hello I am Carlos")
```

```python
label_pipeline("10")
```
