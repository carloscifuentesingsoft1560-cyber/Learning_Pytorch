```python
import torch
import torch.nn as nn
```

```python
class TextClasifire(nn.Module):
    def __init__(self, vocab_size, embedding_dim, hidden_dim, output_dim):
      super().__init__()

      self.embedding = nn.Embedding(vocab_size, embedding_dim)
      self.rnn = nn.LSTM(embedding_dim, hidden_dim, num_layers=2, batch_first=True)
      self.fc = nn.Linear(hidden_dim, output_dim)

      def forward(self, text):
        embedded = self.embedded(text)
        outpu, (hidden,cell) = self.rnn(embedded)
        final_hidden = hidden[-1]
        return self.fc(final_hidden)


```

```python
vocab_size = 10000
embedding_dim = 100
hidden_dim = 256
output_dim = 2

model = TextClasifire(vocab_size,embedding_dim,hidden_dim,output_dim)
```
